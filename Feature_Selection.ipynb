# **Feature Selection Model**

## Strategy: Drop Invalid Data

```python
X = df_kids_DROP_INVALID_FS.drop(columns=['BMI'])
y = df_kids_DROP_INVALID_FS['BMI']
```

```python
le = LabelEncoder()
X_encoded = X.apply(lambda col: le.fit_transform(col.astype(str)))
y_encoded = le.fit_transform(y)

chi2_selector = SelectKBest(score_func=chi2, k='all')
chi2_selector.fit(X_encoded, y_encoded)
```

```python
scores = pd.DataFrame({
    'Feature': X.columns,
    'Chi2 Score': chi2_selector.scores_,
    'p-value': chi2_selector.pvalues_
}).sort_values(by='p-value')

scores['Significant (p ≤ 0.05)'] = scores['p-value'] <= 0.05

print(scores)
```

## Strategy: Replace Invalid Data with Mode

```python
X = df_kids_RP_MODE_FS.drop(columns=['BMI'])
y = df_kids_RP_MODE_FS['BMI']
```

```python
le = LabelEncoder()
X_encoded = X.apply(lambda col: le.fit_transform(col.astype(str)))
y_encoded = le.fit_transform(y)

chi2_selector = SelectKBest(score_func=chi2, k='all')
chi2_selector.fit(X_encoded, y_encoded)
```

```python
scores = pd.DataFrame({
    'Feature': X.columns,
    'Chi2 Score': chi2_selector.scores_,
    'p-value': chi2_selector.pvalues_
}).sort_values(by='p-value')

scores['Significant (p ≤ 0.05)'] = scores['p-value'] <= 0.05

print(scores)
```
