```python
#import libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#preset
sns.set_palette("pastel")
```

```python
file_path = "D:/hp/Documents/Practicum/Code/Dataset/kids_price.csv"

# Read the first line to count the number of columns
with open(file_path, 'r', encoding='utf-8') as x:
    ncols = len(x.readline().strip().split(','))

# Load CSV using the correct number of columns
df_kids = pd.read_csv(file_path, usecols=range(0, ncols))

# Display the DataFrame
df_kids.head()
```

```python
# Replace with the full path to your local CSV file
file_path = "D:/hp/Documents/Practicum/Code/Dataset/df_kids_PP.csv"

# Read the first line to count the number of columns
with open(file_path, 'r', encoding='utf-8') as x:
    ncols = len(x.readline().strip().split(','))

# Load CSV using the correct number of columns
df_kids_PP = pd.read_csv(file_path, usecols=range(0, ncols))

# Display the DataFrame
df_kids_PP.head()
```

# Strategy: Drop Invalid Data

```python
df_kids_DROP_INVALID = df_kids[df_kids['PENDAPATAN KELUARGA'] != 'Tiada Maklumat'].copy()
df_kids_DROP_INVALID.drop(columns=['GAJI BAPA', 'GAJI IBU', 'GAJI PENJAGA'], inplace=True)
df_kids_DROP_INVALID.head()
```

```python
df_kids_DROP_INVALID_FS = df_kids_PP[df_kids_PP['PENDAPATAN KELUARGA'] != 'Tiada Maklumat'].copy()
df_kids_DROP_INVALID_FS.drop(columns=['GAJI BAPA', 'GAJI IBU', 'GAJI PENJAGA'], inplace=True)
df_kids_DROP_INVALID_FS.head()
```

```python
export_path = "D:/hp/Documents/Practicum/Code/Dataset/df_kids_DROP_INVALID.csv"
df_kids_DROP_INVALID.to_csv(export_path, index=False)

export_path = "D:/hp/Documents/Practicum/Code/Dataset/df_kids_DROP_INVALID_FS.csv"
df_kids_DROP_INVALID_FS.to_csv(export_path, index=False)
```

# Strategy: Replace Invalid Data with Mode

```python
df_kids_RP_MODE = df_kids.copy()
df_kids_RP_MODE['PENDAPATAN KELUARGA'].replace('Tiada Maklumat', df_kids_RP_MODE['PENDAPATAN KELUARGA'].mode()[0], inplace=True)
df_kids_RP_MODE.drop(columns=['GAJI BAPA', 'GAJI IBU', 'GAJI PENJAGA'], inplace=True)
df_kids_RP_MODE.head()
```

```python
df_kids_RP_MODE_FS = df_kids_PP.copy()
df_kids_RP_MODE_FS['PENDAPATAN KELUARGA'].replace('Tiada Maklumat', df_kids_RP_MODE_FS['PENDAPATAN KELUARGA'].mode()[0], inplace=True)
df_kids_RP_MODE_FS.drop(columns=['GAJI BAPA', 'GAJI IBU', 'GAJI PENJAGA'], inplace=True)
df_kids_RP_MODE_FS.head()
```

```python
export_path = "D:/hp/Documents/Practicum/Code/Dataset/df_kids_RP_MODE.csv"
df_kids_RP_MODE.to_csv(export_path, index=False)

export_path = "D:/hp/Documents/Practicum/Code/Dataset/df_kids_RP_MODE_FS.csv"
df_kids_RP_MODE_FS.to_csv(export_path, index=False)
```
